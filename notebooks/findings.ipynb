```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/funnel.csv')
df.head()
```

```python
plt.figure(figsize=(10, 6))
plt.bar(df["funnel_step"], df["sessions"])
plt.title("Sessions by Funnel Step")
plt.xlabel("Funnel Step")
plt.ylabel("Sessions")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig('../visuals/sessions_by_funnel_step.png')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(
    df["funnel_step"],
    df["overall_conversion_rate_from_start"],
    marker="o"
)
plt.title("Overall Conversion Rate from Session Start")
plt.xlabel("Funnel Step")
plt.ylabel("Overall Conversion Rate (%)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig('../visuals/overall_conversion_rate_from_start.png')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(
    df["funnel_step"],
    df["step_conversion_rate"],
    marker="o"
)
plt.title("Step-by-Step Funnel Conversion Rate")
plt.xlabel("Funnel Step")
plt.ylabel("Step Conversion Rate (%)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("../visuals/step_conversion_rate_line_chart.png", dpi=300)
plt.show()
```
